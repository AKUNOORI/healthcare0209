```python
%sql
CREATE OR REPLACE TABLE healthcare0209.healthcare0209_gold.patients AS
WITH current_month AS (
    SELECT COUNT(DISTINCT patient_id) AS current_count
    FROM healthcare0209.healthcare0209_silver.encounters
    WHERE admission_date >= '2024-05-01'
      AND admission_date < '2024-06-01'
),

prev_month AS (
    SELECT COUNT(DISTINCT patient_id) AS prev_count
    FROM healthcare0209.healthcare0209_silver.encounters
    WHERE admission_date >= '2024-04-01'
      AND admission_date < '2024-05-01'
)

SELECT
    c.current_count AS Total_patients,
        ROUND(
            ((c.current_count - p.prev_count) * 100.0)
            / NULLIF(p.prev_count, 0),
            2
        )
        AS Patients_increased
FROM current_month c
CROSS JOIN prev_month p;
```

```python
%sql
select * from healthcare0209.healthcare0209_gold.patients
```

```python
%sql
CREATE OR REPLACE TABLE healthcare0209.healthcare0209_gold.encounters_kpi AS

WITH current_month AS (
    SELECT
        COUNT(encounter_id) AS current_count
    FROM healthcare0209.healthcare0209_silver.encounters
    WHERE admission_date >= '2024-05-01'
      AND admission_date < '2024-06-01'
),

prev_month AS (
    SELECT
        COUNT(encounter_id) AS prev_count
    FROM healthcare0209.healthcare0209_silver.encounters
    WHERE admission_date >= '2024-04-01'
      AND admission_date < '2024-05-01'
)

SELECT
    c.current_count AS Total_Encounters,

        ROUND(
            (
                (c.current_count - p.prev_count) * 100.0
                / NULLIF(p.prev_count, 0)
            ),
            2
        )
        AS Encounters_Change

FROM current_month c
CROSS JOIN prev_month p;
```

```python
%sql
select * from healthcare0209.healthcare0209_gold.encounters_kpi
```
