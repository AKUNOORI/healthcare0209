```python
%sql
create or replace table healthcare0209.healthcare0209_gold.allTotalValues
using delta 
as
select count(distinct p.patient_id) as Total_Patients, count(e.encounter_id) as Total_Encounters,
sum(b.paid_amount) as total_revenue
from healthcare0209.healthcare0209_silver.patients p
inner join healthcare0209.healthcare0209_silver.encounters e 
on p.patient_id = e.patient_id
inner join healthcare0209.healthcare0209_silver.billing b 
on e.patient_id = b.patient_id
```
